## Chapter 10 — Cross Products
Concepts: perpendicular vectors, parallelogram area, orientation, and cross products versus dot products

The cross product is a special operation for three-dimensional vectors. It produces a vector perpendicular to both input vectors. The magnitude of that vector equals the area of the parallelogram formed by the inputs.

Unlike the dot product, which produces a scalar, the cross product produces a direction. Reversing the order reverses the result: **a × b = −(b × a)**.

**AI connection:**
Cross products are useful in 3D computer graphics, surface-normal calculations, physics simulations, and robotics. Embedding search uses dot products instead because it needs one scalar similarity score for ranking, not a perpendicular direction vector.

```python
import numpy as np

# --- CROSS PRODUCT ---
# Only works in 3D
# Result is a vector PERPENDICULAR to both input vectors
# Magnitude = area of parallelogram formed by the two vectors
+
+    a = np.array([1, 2, 3])
+    b = np.array([4, 5, 6])
+
+    cross = np.cross(a, b)
+    print(f"a × b = {cross}")
+
+    # --- VERIFY PERPENDICULARITY ---
+    # Cross product should be perpendicular to both a and b
+    # Dot product with either = 0
+    print(f"\na · (a×b) = {np.dot(a, cross)}")  # should be 0
+    print(f"b · (a×b) = {np.dot(b, cross)}")    # should be 0
+
+    # --- MAGNITUDE = AREA ---
+    magnitude = np.linalg.norm(cross)
+    print(f"\n|a × b| = {magnitude:.4f}")
+    print("This equals the area of the parallelogram formed by a and b")
+
+    # --- ORDER MATTERS ---
+    # a × b = -(b × a)
+    cross_ba = np.cross(b, a)
+    print(f"\na × b = {cross}")
+    print(f"b × a = {cross_ba}")
+    print("They are opposite directions — order matters")
+
+    # --- AI CONNECTION ---
+    # Cross products are less common in ML than dot products
+    # But they appear in:
+    # 1. 3D computer vision and graphics
+    # 2. Calculating surface normals in 3D models
+    # 3. Physics simulations used in robotics AI
+
+    # More relevant to you: understanding WHY dot product is used
+    # instead of cross product in embeddings
+    # Dot product → scalar (similarity score) ✅
+    # Cross product → vector (perpendicular direction) ❌ not useful for similarity
+
+    print("\n=== WHY DOT PRODUCT FOR EMBEDDINGS ===")
+    print("Dot product gives a SCALAR — one number = similarity score")
+    print("Cross product gives a VECTOR — not useful for ranking similarity")
+    print("This is why cosine similarity uses dot product, not cross product")
+
+    # Demonstrate
+    king = np.array([0.9, 0.8, 0.1])
+    queen = np.array([0.85, 0.75, 0.9])
+
+    similarity = np.dot(king, queen)
+    print(f"\nking · queen = {similarity:.4f} (scalar — useful)")
+
+    perpendicular = np.cross(king, queen)
+    print(f"king × queen = {perpendicular} (vector — not useful for similarity)")
```

### Key takeaway

- The cross product is defined for 3D vectors and returns a perpendicular vector.
- Its magnitude is the area of the parallelogram formed by the input vectors.
- Swapping the inputs reverses the direction of the result.
- Dot products are better for embedding similarity because they produce scalar scores that can be ranked.
