## Chapter 2 — Linear Combinations, Span, Basis Vectors
Watched: 3Blue1Brown Essence of LA, Video 2
Concepts: linear combinations, span, basis vectors, linear dependence

**AI connection:** 
- Every embedding is a linear combination of basis dimensions
- The "span" of all possible embeddings = the model's entire semantic understanding
- Normalized embeddings (magnitude = 1) enable cosine similarity

```python
import numpy as np

# --- LINEAR COMBINATIONS ---
# Any vector can be made by scaling and adding basis vectors
i_hat = np.array([1, 0])  # standard basis vector x
j_hat = np.array([0, 1])  # standard basis vector y

# Vector [3, -2] is just 3*i_hat + (-2)*j_hat
v = 3 * i_hat + (-2) * j_hat
print("Linear combination result:", v)  # [3, -2]

# --- SPAN ---
# The span of two vectors is every possible linear combination of them
# If two vectors point in different directions — span is all of 2D space
# If they're parallel — span is just a line

# Visualize: generate 10 random linear combinations
import matplotlib.pyplot as plt

points = []
for a in np.linspace(-2, 2, 10):
    for b in np.linspace(-2, 2, 10):
        point = a * i_hat + b * j_hat
        points.append(point)

points = np.array(points)
plt.scatter(points[:, 0], points[:, 1], s=10)
plt.title("Span of i_hat and j_hat — covers all 2D space")
plt.grid(True)
plt.show()

# --- LINEARLY DEPENDENT (parallel vectors) ---
v1 = np.array([1, 2])
v2 = np.array([2, 4])  # v2 is just 2*v1 — same direction
# These two vectors span only a LINE, not 2D space

# --- AI CONNECTION ---
# In a 1536-dimension embedding space:
# Each dimension is like a basis vector
# An embedding is a linear combination of all 1536 basis directions
# "Span" of all embeddings = the entire semantic space the model understands

print("\nEmbedding dimension example:")
embedding_dim = 1536
random_embedding = np.random.randn(embedding_dim)
normalized = random_embedding / np.linalg.norm(random_embedding)
print(f"Embedding shape: {normalized.shape}")
print(f"Magnitude after normalization: {np.linalg.norm(normalized):.4f}")
# Always 1.0 — this is why normalized vectors work for cosine similarity
```
