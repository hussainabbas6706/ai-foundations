## Chapter 8 — Non-Square Matrices
Concepts: changing dimensions, information loss, neural-network layers, and embeddings

A non-square matrix has a different number of rows and columns. The columns represent input dimensions, while the rows represent output dimensions. This allows a matrix to transform vectors from one-dimensional space into another.

A matrix can expand a vector into a higher-dimensional space or compress it into a lower-dimensional space. Expansion adds dimensions but does not automatically add information. Compression usually loses information because multiple input vectors can produce the same output.

**AI connection:**
Neural-network layers are often non-square matrix transformations. They change the width of a representation, such as mapping token space into an embedding space or mapping hidden activations to model outputs.

```python
import numpy as np

# --- NON-SQUARE MATRICES ---
# Rows = output dimensions
# Columns = input dimensions

# --- 3x2 MATRIX: 2D → 3D ---
# Takes a 2D vector and lifts it into 3D space
matrix_3x2 = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

input_2d = np.array([1, 2])
output_3d = matrix_3x2 @ input_2d

print("=== 2D → 3D TRANSFORMATION ===")
print(f"Input shape: {input_2d.shape}")    # (2,)
print(f"Matrix shape: {matrix_3x2.shape}") # (3, 2)
print(f"Output shape: {output_3d.shape}")  # (3,)
print(f"Input vector: {input_2d}")
print(f"Output vector: {output_3d}")

# --- 2x3 MATRIX: 3D → 2D ---
# Takes a 3D vector and compresses it into 2D space
matrix_2x3 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

input_3d = np.array([1, 2, 3])
output_2d = matrix_2x3 @ input_3d

print("\n=== 3D → 2D TRANSFORMATION ===")
print(f"Input shape: {input_3d.shape}")    # (3,)
print(f"Matrix shape: {matrix_2x3.shape}") # (2, 3)
print(f"Output shape: {output_2d.shape}")  # (2,)
print(f"Input vector: {input_3d}")
print(f"Output vector: {output_2d}")

# --- INFORMATION LOSS ---
# Going from high dimension to low dimension loses information
# Going from low dimension to high dimension adds dimensions but not information

print("\n=== INFORMATION LOSS CHECK ===")
# 3D → 2D loses one dimension
rank_compress = np.linalg.matrix_rank(matrix_2x3)
print(f"Compression matrix rank: {rank_compress}")
print("Information is lost when compressing — cannot fully reverse this")

# --- NEURAL NETWORK LAYER SIMULATION ---
print("\n=== NEURAL NETWORK LAYERS ===")

# Layer 1: 3 inputs → 5 hidden neurons (3D → 5D)
W1 = np.random.randn(5, 3)
# Layer 2: 5 hidden → 4 hidden neurons (5D → 4D)
W2 = np.random.randn(4, 5)
# Layer 3: 4 hidden → 2 outputs (4D → 2D)
W3 = np.random.randn(2, 4)

input_vec = np.array([0.5, 0.3, 0.8])

layer1_out = W1 @ input_vec
layer2_out = W2 @ layer1_out
layer3_out = W3 @ layer2_out

print(f"Input:   {input_vec.shape}")    # (3,)
print(f"Layer 1: {layer1_out.shape}")   # (5,)
print(f"Layer 2: {layer2_out.shape}")   # (4,)
print(f"Layer 3: {layer3_out.shape}")   # (2,)
print("\nEach layer is a non-square matrix transformation")

# --- EMBEDDING CONNECTION ---
print("\n=== EMBEDDING MODEL SIMULATION ===")
vocab_size = 50000   # 50k possible tokens
embed_dim = 3072     # Gemini embedding dimensions

# This is what happens inside an embedding model
# Non-square transformation: vocab_size → embed_dim
W_embed = np.random.randn(embed_dim, vocab_size)

# One-hot vector for the word "king" (token index 1337)
king_token = np.zeros(vocab_size)
king_token[1337] = 1.0

king_embedding = W_embed @ king_token
print(f"Token input shape:    {king_token.shape}")      # (50000,)
print(f"Embedding matrix:     {W_embed.shape}")          # (3072, 50000)
print(f"Embedding output:     {king_embedding.shape}")   # (3072,)
print("\nThis is exactly what happens when you call:")
print("client.models.embed_content(model='gemini-embedding-001', contents='king')")
print("50,000 token space → 3,072 dimensional embedding space")
```

### Key takeaway

- Rows describe output dimensions, and columns describe input dimensions.
- Non-square matrices can expand or compress vector representations.
- Compression can lose information because the transformation is not reversible in general.
- Neural-network layers and embedding models use non-square matrices to change representation size.
