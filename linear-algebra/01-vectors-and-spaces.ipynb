# Vectors and Spaces

```python
import numpy as np

# --- VECTORS ---
# A vector is just a list of numbers — a point in space
v1 = np.array([2, 3])
v2 = np.array([1, -1])
print("v1:", v1)
print("v2:", v2)

# --- VECTOR ADDITION ---
# Add tip to tail — combine two movements
result = v1 + v2
print("v1 + v2:", result)

# --- SCALAR MULTIPLICATION ---
# Stretch or shrink a vector — same direction, different magnitude
scaled = 3 * v1
print("3 * v1:", scaled)

# --- NEGATIVE SCALAR — reverses direction ---
flipped = -1 * v1
print("-v1:", flipped)

# --- THE AI CONNECTION ---
# In embeddings, a word like "king" is stored as a vector
# e.g. king = [0.9, 0.8, 0.1, ...]
# Vector addition is how we do "king - man + woman = queen"
king =   np.array([0.9, 0.8, 0.1])
man =    np.array([0.6, 0.5, 0.1])
woman =  np.array([0.5, 0.4, 0.9])

result = king - man + woman
print("king - man + woman:", result)
# This result should be close to the queen vector
queen =  np.array([0.85, 0.75, 0.9])
print("queen:             ", queen)
```
