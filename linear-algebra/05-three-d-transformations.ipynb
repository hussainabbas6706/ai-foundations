## Chapter 5 — 3D Transformations
Watched: 3Blue1Brown Essence of LA, Video 5
Concepts: 3D matrices, rotation in 3D, extending 2D intuition to higher dimensions

**AI connection:**
Embeddings are 1536-dimensional vectors.
Attention layers in LLMs are 1536x1536 matrix transformations.
Same math as 2D rotation — just more dimensions.
The intuition from 2D and 3D directly applies to 1536D.

```python
import numpy as np

# --- 3D TRANSFORMATIONS ---
# Same concept as 2D but with 3x3 matrices

# 3D basis vectors
i_hat = np.array([1, 0, 0])
j_hat = np.array([0, 1, 0])
k_hat = np.array([0, 0, 1])

# Rotation around Z axis by 90 degrees
rotation_z = np.array([
    [0, -1, 0],
    [1,  0, 0],
    [0,  0, 1]
])

# Where do basis vectors land?
print("i_hat after rotation:", rotation_z @ i_hat)
print("j_hat after rotation:", rotation_z @ j_hat)
print("k_hat after rotation:", rotation_z @ k_hat)

# --- APPLY TO ANY 3D VECTOR ---
v = np.array([1, 2, 3])
print("\nv transformed:", rotation_z @ v)

# --- SCALING IN 3D ---
scale_3d = np.array([
    [2, 0, 0],
    [0, 2, 0],
    [0, 0, 2]
])
print("v scaled 2x:", scale_3d @ v)

# --- AI CONNECTION ---
# Embeddings are NOT 2D or 3D — they are 1536D
# But the exact same rules apply
# Every transformer attention operation is a high-dimensional transformation

embedding = np.random.randn(1536)  # a real OpenAI embedding size
weight_matrix = np.random.randn(1536, 1536)  # attention weight matrix
transformed = weight_matrix @ embedding
print("\n1536D embedding transformation:")
print("Input shape:", embedding.shape)
print("Output shape:", transformed.shape)
# This is one attention head in an LLM — just a 1536x1536 matrix multiplication
```
