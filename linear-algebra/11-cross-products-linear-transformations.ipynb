## Chapter 11 — Cross Products as Linear Transformations
Concepts: cross-product transformations, skew-symmetric matrices, and geometric meaning

For a fixed vector **a**, taking the cross product with another vector **v** can be viewed as a linear transformation:

**T(v) = a × v**

Although the cross product is not linear in both vectors at the same time, it is linear with respect to one vector when the other is held fixed. This means that the operation can be represented by a matrix.

For **a = [a₁, a₂, a₃]**, the cross product **a × v** can be written using the skew-symmetric matrix:

```text
[  0  -a₃   a₂ ]
[ a₃   0   -a₁ ]  v
[-a₂  a₁    0  ]
```

This matrix always maps vectors into directions perpendicular to **a**. The vector parallel to **a** is mapped to zero, so the transformation has a non-trivial null space and cannot be invertible.

**AI and graphics connection:**
Cross-product transformations are important in 3D graphics, robotics, physics simulations, and computer vision. They help calculate surface normals, rotations, torques, and geometric orientations. This is different from embedding similarity, where dot products produce scalar scores for ranking.
