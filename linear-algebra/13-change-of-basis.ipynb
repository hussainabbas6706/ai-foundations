## Chapter 13 — Change of Basis
Concepts: coordinate systems, basis matrices, transformed operators, and attention projections

A vector is an abstract object, while its coordinates depend on the basis used to describe it. Changing the basis changes the coordinates, not the underlying vector.

If the columns of **B** contain the vectors of a new basis, then coordinates in the new basis can be converted into standard coordinates with:

**v_standard = B v_new**

To convert back, use the inverse basis matrix:

**v_new = B⁻¹ v_standard**

A transformation can also be expressed in a different basis. If **T** is the transformation in standard coordinates, then its representation in the new basis is:

**T_new = B⁻¹ T B**

**AI connection:**
Transformer attention uses learned projection matrices to view the same embedding in different representation spaces. Query, key, and value projections act like learned changes of coordinates, allowing each attention head to focus on different relationships.

```python
import numpy as np

# --- CHANGE OF BASIS ---
# Same vector, different coordinate systems
# Like describing Karachi's location in GPS coords vs local landmarks
+
+    # Our standard basis
+    i_hat = np.array([1, 0])
+    j_hat = np.array([0, 1])
+
+    # Jennifer's basis (different coordinate system)
+    b1 = np.array([2, 1])   # Jennifer's x-axis
+    b2 = np.array([-1, 1])  # Jennifer's y-axis
+
+    # Basis change matrix — columns are Jennifer's basis vectors
+    B = np.column_stack([b1, b2])
+    print("Jennifer's basis matrix:")
+    print(B)
+
+    # --- OUR LANGUAGE → JENNIFER'S LANGUAGE ---
+    # Jennifer says vector is [3, 2] in her system
+    jennifer_coords = np.array([3, 2])
+
+    # Convert to our system
+    our_coords = B @ jennifer_coords
+    print(f"\nJennifer's coords: {jennifer_coords}")
+    print(f"Our coords:        {our_coords}")
+
+    # --- JENNIFER'S LANGUAGE → OUR LANGUAGE ---
+    B_inv = np.linalg.inv(B)
+    back = B_inv @ our_coords
+    print(f"Back to Jennifer:  {np.round(back, 2)}")
+
+    # --- TRANSFORMING IN JENNIFER'S SYSTEM ---
+    # Apply a transformation in our system
+    # but express it in Jennifer's coordinate system
+
+    # Rotation 90 degrees in our system
+    rotation = np.array([
+        [0, -1],
+        [1,  0]
+    ])
+
+    # Same rotation expressed in Jennifer's language
+    # B_inv @ rotation @ B
+    jennifer_rotation = B_inv @ rotation @ B
+    print("\nRotation in our system:")
+    print(rotation)
+    print("\nSame rotation in Jennifer's system:")
+    print(np.round(jennifer_rotation, 2))
+
+    # --- AI CONNECTION ---
+    # This is exactly what happens in transformer attention
+    # Each attention head operates in a different "basis"
+    # Q, K, V projections = change of basis operations
+
+    print("\n=== TRANSFORMER ATTENTION CONNECTION ===")
+    d_model = 512   # embedding dimension
+    d_k = 64        # key/query dimension per head
+
+    # Input embedding
+    x = np.random.randn(d_model)
+
+    # Weight matrices = basis change matrices
+    W_q = np.random.randn(d_k, d_model)  # change to query basis
+    W_k = np.random.randn(d_k, d_model)  # change to key basis
+    W_v = np.random.randn(d_k, d_model)  # change to value basis
+
+    Q = W_q @ x  # x in query basis
+    K = W_k @ x  # x in key basis
+    V = W_v @ x  # x in value basis
+
+    print(f"Input:  {x.shape}")
+    print(f"Query:  {Q.shape} — input in query basis")
+    print(f"Key:    {K.shape} — input in key basis")
+    print(f"Value:  {V.shape} — input in value basis")
+    print("\nEach head sees the same input from a different basis perspective")
+    print("This is how attention captures different types of relationships")
```

### Key takeaway

- Coordinates describe a vector relative to a chosen basis.
- A basis matrix converts coordinates between coordinate systems.
- The expression **B⁻¹TB** represents the same transformation in a new basis.
- Transformer projections create learned representation spaces for queries, keys, and values.
