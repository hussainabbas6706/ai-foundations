## Chapter 9 — Dot Products and Duality
Concepts: geometric meaning, algebraic dot products, projections, cosine similarity, and RAG retrieval

The dot product measures how much two vectors point in the same direction. Algebraically, it is the sum of element-wise products. Geometrically, it is the product of the vectors' magnitudes and the cosine of the angle between them.

A positive dot product indicates that vectors point generally in the same direction, zero indicates perpendicular vectors, and a negative value indicates opposing directions.

**Duality:**
A vector can also act as a transformation. Treating a vector as a 1 × N matrix lets it project another vector onto its direction and produce a scalar score.

**AI connection:**
Cosine similarity is a normalized dot product. Embedding search compares a query vector with stored document vectors and returns the documents with the highest similarity scores. This is the mathematical foundation of retrieval-augmented generation (RAG).

```python
import numpy as np

# --- DOT PRODUCT ---
# Two ways to think about it:
# 1. Geometric: |a| * |b| * cos(angle between them)
# 2. Algebraic: sum of element-wise products

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Algebraic dot product
dot_algebraic = np.dot(a, b)
print(f"Dot product (algebraic): {dot_algebraic}")

# Geometric dot product
magnitude_a = np.linalg.norm(a)
magnitude_b = np.linalg.norm(b)
cos_angle = dot_algebraic / (magnitude_a * magnitude_b)
angle = np.arccos(cos_angle) * 180 / np.pi
print(f"Angle between vectors: {angle:.2f} degrees")

# --- WHAT DOT PRODUCT TELLS YOU ---
print("\n=== WHAT DOT PRODUCT MEANS ===")
# Positive = vectors point in same direction
# Zero = vectors are perpendicular (unrelated)
# Negative = vectors point in opposite directions

v1 = np.array([1, 0])    # pointing right
v2 = np.array([1, 0])    # same direction
v3 = np.array([0, 1])    # perpendicular
v4 = np.array([-1, 0])   # opposite direction

print(f"Same direction:     {np.dot(v1, v2)}")   # positive
print(f"Perpendicular:      {np.dot(v1, v3)}")   # zero
print(f"Opposite direction: {np.dot(v1, v4)}")   # negative

# --- DUALITY ---
# A vector can act as a transformation (1xN matrix)
# It projects other vectors onto itself
w = np.array([2, 1])

# w as a transformation — projects any vector onto w's direction
vectors = [np.array([3, 1]), np.array([1, 3]), np.array([-1, 2])]
print("\n=== PROJECTION VIA DUALITY ===")
for v in vectors:
    projection = np.dot(w, v)
    print(f"w · {v} = {projection}")

# --- COSINE SIMILARITY = NORMALIZED DOT PRODUCT ---
print("\n=== COSINE SIMILARITY ===")
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

# This is exactly what embedding search does
king = np.array([0.9, 0.8, 0.1])
queen = np.array([0.85, 0.75, 0.9])
biryani = np.array([0.1, 0.1, 0.8])

print(f"king vs queen:   {cosine_similarity(king, queen):.4f}")
print(f"king vs biryani: {cosine_similarity(king, biryani):.4f}")

# --- AI CONNECTION ---
print("\n=== AI CONNECTION ===")
print("Cosine similarity IS normalized dot product")
print("Every embedding search you do:")
print("1. Embeds your query → vector")
print("2. Computes dot product with every stored vector")
print("3. Normalizes by magnitudes = cosine similarity")
print("4. Returns highest scores = most similar documents")
print("\nThis is the entire mathematical foundation of RAG retrieval")

# Simulate RAG retrieval
query = np.array([0.8, 0.7, 0.2])
documents = {
    "Python tutorial": np.array([0.75, 0.65, 0.15]),
    "Biryani recipe": np.array([0.1, 0.05, 0.9]),
    "AI engineering": np.array([0.85, 0.8, 0.1]),
    "Cricket match": np.array([0.2, 0.1, 0.6]),
}

print("\n=== RAG RETRIEVAL SIMULATION ===")
results = []
for doc, vec in documents.items():
    score = cosine_similarity(query, vec)
    results.append((score, doc))

results.sort(reverse=True)
for score, doc in results:
    print(f"{score:.4f} — {doc}")
```

### Key takeaway

- The dot product measures directional alignment between vectors.
- A vector can act as a linear transformation through duality.
- Cosine similarity removes magnitude and focuses on direction.
- RAG retrieval ranks documents by comparing the query embedding with stored embeddings.
