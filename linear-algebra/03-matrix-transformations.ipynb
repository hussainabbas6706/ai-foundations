## Chapter 3 — Matrix Transformations
Watched: 3Blue1Brown Essence of LA, Video 3
Concepts: matrices as transformations, basis vectors after transformation,
rotation, scaling, shear

**AI connection:**
Every layer in a neural network is a matrix transformation.
Input embeddings get multiplied by weight matrices at every layer.
This is how LLMs process and transform meaning through their layers.

```python
import numpy as np
import matplotlib.pyplot as plt

# --- MATRIX AS A TRANSFORMATION ---
# A matrix transforms space — it moves every vector in a predictable way

# Original basis vectors
i_hat = np.array([1, 0])
j_hat = np.array([0, 1])

# Rotation 90 degrees matrix
rotation = np.array([
    [0, -1],
    [1,  0]
])

# Where do basis vectors land after transformation?
new_i = rotation @ i_hat
new_j = rotation @ j_hat
print("i_hat after rotation:", new_i)  # [0, 1]
print("j_hat after rotation:", new_j)  # [-1, 0]

# --- APPLY TRANSFORMATION TO ANY VECTOR ---
v = np.array([3, 2])
transformed = rotation @ v
print("v after rotation:", transformed)

# --- SCALING TRANSFORMATION ---
scale = np.array([
    [2, 0],
    [0, 2]
])
print("v scaled 2x:", scale @ v)

# --- SHEAR TRANSFORMATION ---
shear = np.array([
    [1, 1],
    [0, 1]
])
print("v sheared:", shear @ v)

# --- AI CONNECTION ---
# Every layer in a neural network is a matrix transformation
# Input vector → multiply by weight matrix → transformed vector
# This is how neural networks learn — they learn the right transformation matrices

input_vector = np.array([0.5, 0.3, 0.8])
weight_matrix = np.random.randn(4, 3)  # transforms 3D → 4D
output = weight_matrix @ input_vector
print("
Neural network layer simulation:")
print("Input shape:", input_vector.shape)
print("Weight matrix shape:", weight_matrix.shape)
print("Output shape:", output.shape)
# This is literally what happens in every LLM layer
```
