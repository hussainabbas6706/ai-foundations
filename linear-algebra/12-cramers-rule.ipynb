## Chapter 12 — Cramer's Rule
Concepts: solving linear systems with determinants, geometric interpretation, and least-squares regression

Cramer's Rule solves a square system of linear equations by replacing one column of the coefficient matrix with the output vector and comparing determinants.

For a system **Ax = b**, each unknown is found by a ratio of determinants. For example, the first component is:

**x₁ = det(A₁) / det(A)**

where **A₁** is formed by replacing the first column of **A** with **b**. Cramer's Rule requires **det(A) ≠ 0**, because a singular matrix does not have a unique solution.

**Geometric interpretation:**
The determinant measures the area of a parallelogram. Each solution component is the ratio between the area after replacing a column with **b** and the original area.

**AI connection:**
Cramer's Rule is rarely used directly in large machine-learning systems, but solving **Ax = b** is fundamental to linear regression, least-squares optimization, and finding unknown model parameters.

```python
import numpy as np

# --- CRAMER'S RULE ---
# Solving a system of linear equations geometrically
# Ax = b → find x using determinants
+
+    # System of equations:
+    # 3x + 1y = 5
+    # 1x + 2y = 4
+
+    A = np.array([
+        [3, 1],
+        [1, 2]
+    ], dtype=float)
+
+    b = np.array([5, 4], dtype=float)
+
+    # --- METHOD 1: CRAMER'S RULE ---
+    det_A = np.linalg.det(A)
+    print(f"det(A) = {det_A:.2f}")
+
+    # For x: replace first column with b
+    A_x = A.copy()
+    A_x[:, 0] = b
+    x = np.linalg.det(A_x) / det_A
+
+    # For y: replace second column with b
+    A_y = A.copy()
+    A_y[:, 1] = b
+    y = np.linalg.det(A_y) / det_A
+
+    print("\nCramer's Rule solution:")
+    print(f"x = {x:.2f}")
+    print(f"y = {y:.2f}")
+
+    # --- METHOD 2: NUMPY BUILT-IN ---
+    solution = np.linalg.solve(A, b)
+    print("\nNumPy solution:")
+    print(f"x = {solution[0]:.2f}")
+    print(f"y = {solution[1]:.2f}")
+    print("Same result ✅")
+
+    # --- VERIFY ---
+    print("\nVerify A @ solution = b:")
+    print(np.round(A @ solution, 2))
+
+    # --- GEOMETRIC INTERPRETATION ---
+    # det(A) = area of parallelogram formed by columns of A
+    # det(A_x) = area when column replaced by b
+    # x = ratio of these areas
+    print("\nGeometric interpretation:")
+    print(f"det(A) = {det_A:.2f} — area of original parallelogram")
+    print(f"det(A_x) = {np.linalg.det(A_x):.2f} — area after replacing column")
+    print(f"x = {np.linalg.det(A_x):.2f} / {det_A:.2f} = {x:.2f}")
+
+    # --- AI CONNECTION ---
+    # Cramer's rule is rarely used directly in AI
+    # But the concept of solving Ax = b is everywhere:
+    # - Linear regression: finds weights that minimize error
+    # - Attention mechanism: solves for optimal attention weights
+    # - Any optimization problem in ML
+
+    print("\n=== AI CONNECTION ===")
+    print("Solving Ax = b is the foundation of:")
+    print("- Linear regression (finding optimal weights)")
+    print("- Least squares optimization")
+    print("- Any system where you need to find unknown parameters")
+
+    # Linear regression example
+    # Given data points, find best fit line y = mx + c
+    X_data = np.array([[1, 1], [2, 1], [3, 1], [4, 1]], dtype=float)
+    y_data = np.array([2, 4, 5, 4], dtype=float)
+
+    # Solve using least squares (Ax = b under the hood)
+    coeffs, _, _, _ = np.linalg.lstsq(X_data, y_data, rcond=None)
+    print(f"\nLinear regression coefficients: slope={coeffs[0]:.2f}, intercept={coeffs[1]:.2f}")
+    print("This solved Ax = b using the same principle as Cramer's rule")
```

### Key takeaway

- Cramer's Rule uses determinant ratios to solve square linear systems.
- A unique solution requires a non-zero determinant.
- Determinant ratios provide a geometric interpretation of each solution component.
- The broader idea of solving linear systems supports regression and optimization in machine learning.
