## Chapter 7 — Inverse Matrices, Column Space, and Null Space
Concepts: matrix inverses, solving linear systems, rank, column space, and null space

An inverse matrix reverses an invertible transformation. The column space describes every output a matrix can produce, while the null space contains directions that the matrix maps to zero.

**AI connection:**
Rank measures how much information a matrix preserves. Low-rank transformations create information bottlenecks and dimensionality reduction, which are central ideas behind PCA and efficient neural-network representations.

```python
import numpy as np

# --- INVERSE MATRIX ---
# A_inv such that A @ A_inv = Identity matrix
A = np.array([
    [3, 1],
    [1, 2]
], dtype=float)

A_inv = np.linalg.inv(A)
print("A inverse:")
print(A_inv)

# Verify: A @ A_inv should = Identity
identity = A @ A_inv
print("\nA @ A_inv (should be identity):")
print(np.round(identity, 2))

# --- SOLVING LINEAR SYSTEMS ---
# Ax = b → x = A_inv @ b
b = np.array([5, 3], dtype=float)
x = A_inv @ b
print(f"\nSolution x: {x}")
print(f"Verify A @ x = b: {np.round(A @ x, 2)}")

# --- COLUMN SPACE ---
# All possible outputs of a matrix transformation
# = span of its column vectors
M = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])
rank = np.linalg.matrix_rank(M)
print(f"\nRank of M (dimension of column space): {rank}")

# --- NULL SPACE ---
# All vectors that get squished to zero by the transformation
# If null space has only zero vector — transformation is invertible
singular = np.array([
    [1, 2],
    [2, 4]  # row 2 = 2 * row 1 — linearly dependent
])
print(f"Rank of singular matrix: {np.linalg.matrix_rank(singular)}")
print(f"Null space exists: {np.linalg.matrix_rank(singular) < singular.shape[1]}")

# --- AI CONNECTION ---
# Rank tells you how much information a matrix preserves
# Low rank = information bottleneck = dimensionality reduction
# This is the foundation of PCA and attention mechanisms in LLMs
# Null space = directions where the model loses information completely

print("\n=== EMBEDDING SPACE RANK ===")
# Simulate checking if embedding matrix preserves information
embedding_matrix = np.random.randn(10, 768)  # 10 embeddings, 768 dims
rank = np.linalg.matrix_rank(embedding_matrix)
print(f"Embedding matrix rank: {rank} out of {min(embedding_matrix.shape)}")
print("Full rank = no information lost in this embedding space")
```

### Key takeaway

- An inverse reverses an invertible matrix transformation.
- The column space is the set of all outputs a matrix can produce.
- The null space contains directions that are completely lost.
- Rank measures the number of independent directions preserved by a matrix.
