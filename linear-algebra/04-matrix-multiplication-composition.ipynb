## Chapter 4 — Matrix Multiplication as Composition
Watched: 3Blue1Brown Essence of LA, Video 4
Concepts: composition of transformations, order matters, matrix multiplication

**AI connection:**
A neural network forward pass is just chained matrix multiplications.
W3 @ W2 @ W1 @ input = the entire forward pass of a 3-layer network.
LLMs have hundreds of these layers — each one a learned transformation.

```python
import numpy as np

# --- MATRIX MULTIPLICATION AS COMPOSITION ---
# Applying two transformations one after another

# Rotation 90 degrees
rotation = np.array([
    [0, -1],
    [1,  0]
])

# Shear transformation
shear = np.array([
    [1, 1],
    [0, 1]
])

# Applying shear THEN rotation = multiply rotation @ shear
# Order matters — right to left
composed = rotation @ shear
print("Composed transformation matrix:")
print(composed)

# --- VERIFY ---
# Apply both separately vs composed — should be identical
v = np.array([1, 2])

step1 = shear @ v        # shear first
step2 = rotation @ step1 # then rotate
print("\nApplied separately:", step2)

combined = composed @ v  # both at once
print("Applied composed:  ", combined)
# Same result — that's what matrix multiplication means

# --- ORDER MATTERS ---
# rotation @ shear ≠ shear @ rotation
reversed_order = shear @ rotation
print("\nReversed order matrix:")
print(reversed_order)
print("Different from:", composed)

# --- AI CONNECTION ---
# In a neural network with multiple layers:
# Layer1_weights @ Layer2_weights @ Layer3_weights @ input
# Each @ is a composition of transformations
# The network learns the right sequence of transformations
# to map raw input → meaningful output

# Simulating 3 layer neural network composition
W1 = np.random.randn(4, 3)  # layer 1: 3D → 4D
W2 = np.random.randn(5, 4)  # layer 2: 4D → 5D
W3 = np.random.randn(2, 5)  # layer 3: 5D → 2D

input_vec = np.array([0.5, 0.3, 0.8])

# Sequential application
out = W3 @ W2 @ W1 @ input_vec
print("\n3-layer network output shape:", out.shape)
print("Output:", out)
# This is the forward pass of a neural network — just matrix multiplication
```
